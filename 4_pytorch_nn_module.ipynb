```python
#importing the required libraries
import torch
import torch.nn as nn
```

```python
#create a model class -
class Model(nn.Module): #class have to inherit from nn.Module to use offerings of nn.Modules
  def __init__(self,Num_features): #constructor

    super().__init__() #invoke custructor of parent class
    self.linear = nn.Linear(Num_features, 1)     #creating Layer(have to pass number of inputs features and output)
    self.sigmoid = nn.Sigmoid() #activation funtion

  #Forward pass
  def forward(self, features):
    #calculate the output when u pass input values to the NN.
    out = self.linear(features) #calulating  Z= wx+b storing in out
    out = self.sigmoid(out)

    return out
```

```python
#create dataset
features = torch.rand(10,5)

#create model
model = Model(features.shape[1])

#call model for foward pass
model(features) #this is standard way of calling in pytorch instead of #model.forward(features)
```
